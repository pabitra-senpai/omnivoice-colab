## 🛠️ Troubleshooting

- **`ModuleNotFoundError: omnivoice`** → Re-run cell #2 and check whether `colab.txt` finished installing correctly (if the `assert` above fails, it will show a detailed error).
- **CUDA Out of Memory** → Lower Inference Steps in the UI (32 → 16), or Runtime → Restart session and start again.
- **Gradio link not working** → Stop the cell and re-run it — you'll get a new link.
- **Full language list** → [docs/languages.md](https://github.com/k2-fsa/OmniVoice/blob/master/docs/languages.md)
- **All Voice Design attributes** → [docs/voice-design.md](https://github.com/k2-fsa/OmniVoice/blob/master/docs/voice-design.md)
