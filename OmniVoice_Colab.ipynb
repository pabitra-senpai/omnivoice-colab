## 🛠️ Troubleshooting

- **`ModuleNotFoundError: omnivoice`** → সেল #2 আবার রান করো, `colab.txt` ইনস্টল ঠিকমতো শেষ হয়েছে কিনা দেখো (উপরের `assert` ফেইল করলে বিস্তারিত এরর দেখাবে)।
- **CUDA Out of Memory** → UI-তে Inference Steps কমাও (৩২ → ১৬), অথবা Runtime → Restart session করে আবার শুরু করো।
- **Gradio লিংক কাজ করছে না** → সেলটা Stop করে আবার রান করো, নতুন লিংক পাবে।
- **সব ভাষার তালিকা** → [docs/languages.md](https://github.com/k2-fsa/OmniVoice/blob/master/docs/languages.md)
- **Voice Design এর সব attribute** → [docs/voice-design.md](https://github.com/k2-fsa/OmniVoice/blob/master/docs/voice-design.md)
